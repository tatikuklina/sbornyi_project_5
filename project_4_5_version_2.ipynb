# Сборный проект-4

Вам поручено разработать демонстрационную версию поиска изображений по запросу.

Для демонстрационной версии нужно обучить модель, которая получит векторное представление изображения, векторное представление текста, а на выходе выдаст число от 0 до 1 — покажет, насколько текст и картинка подходят друг другу.

### Описание данных

Данные доступны по [ссылке](https://code.s3.yandex.net/datasets/dsplus_integrated_project_4.zip).

В файле `train_dataset.csv` находится информация, необходимая для обучения: имя файла изображения, идентификатор описания и текст описания. Для одной картинки может быть доступно до 5 описаний. Идентификатор описания имеет формат `<имя файла изображения>#<порядковый номер описания>`.

В папке `train_images` содержатся изображения для тренировки модели.

В файле `CrowdAnnotations.tsv` — данные по соответствию изображения и описания, полученные с помощью краудсорсинга. Номера колонок и соответствующий тип данных:

1. Имя файла изображения.
2. Идентификатор описания.
3. Доля людей, подтвердивших, что описание соответствует изображению.
4. Количество человек, подтвердивших, что описание соответствует изображению.
5. Количество человек, подтвердивших, что описание не соответствует изображению.

В файле `ExpertAnnotations.tsv` содержатся данные по соответствию изображения и описания, полученные в результате опроса экспертов. Номера колонок и соответствующий тип данных:

1. Имя файла изображения.
2. Идентификатор описания.

3, 4, 5 — оценки трёх экспертов.

Эксперты ставят оценки по шкале от 1 до 4, где 1 — изображение и запрос совершенно не соответствуют друг другу, 2 — запрос содержит элементы описания изображения, но в целом запрос тексту не соответствует, 3 — запрос и текст соответствуют с точностью до некоторых деталей, 4 — запрос и текст соответствуют полностью.

В файле `test_queries.csv` находится информация, необходимая для тестирования: идентификатор запроса, текст запроса и релевантное изображение. Для одной картинки может быть доступно до 5 описаний. Идентификатор описания имеет формат `<имя файла изображения>#<порядковый номер описания>`.

В папке `test_images` содержатся изображения для тестирования модели.

## Исследовательский анализ данных

Наш датасет содержит экспертные и краудсорсинговые оценки соответствия текста и изображения.

В файле с экспертными мнениями для каждой пары изображение-текст имеются оценки от трёх специалистов. Для решения задачи вы должны эти оценки агрегировать — превратить в одну. Существует несколько способов агрегации оценок, самый простой — голосование большинства: за какую оценку проголосовала большая часть экспертов (в нашем случае 2 или 3), та оценка и ставится как итоговая. Поскольку число экспертов меньше числа классов, может случиться, что каждый эксперт поставит разные оценки, например: 1, 4, 2. В таком случае данную пару изображение-текст можно исключить из датасета.

Вы можете воспользоваться другим методом агрегации оценок или придумать свой.

В файле с краудсорсинговыми оценками информация расположена в таком порядке:

1. Доля исполнителей, подтвердивших, что текст **соответствует** картинке.
2. Количество исполнителей, подтвердивших, что текст **соответствует** картинке.
3. Количество исполнителей, подтвердивших, что текст **не соответствует** картинке.

После анализа экспертных и краудсорсинговых оценок выберите либо одну из них, либо объедините их в одну по какому-то критерию: например, оценка эксперта принимается с коэффициентом 0.6, а крауда — с коэффициентом 0.4.

Ваша модель должна возвращать на выходе вероятность соответствия изображения тексту, поэтому целевая переменная должна иметь значения от 0 до 1.

### Агрегация данных (ExpertAnnotations)

````python
# функция голосования
def get_majority(row):
    votes = [row['r_1'], row['r_2'], row['r_3']]
    for v in set(votes):
        if votes.count(v) >= 2:
            return v
    return None
````

````python
# использоание функции голосования
df_annotations_expert['final_rating'] = df_annotations_expert.apply(get_majority, axis=1)
````

````python
# Предобработка текста
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

def clean_text(text):
    # Приведение к нижнему регистру
    text = text.lower()
    
    # Удаление URL
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)
    
    # Удаление HTML тегов
    text = re.sub(r'<.*?>', '', text)
    
    # Удаление пунктуации и цифр (оставляем только буквы и пробелы)
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    
    # Удаление лишних пробелов
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

def lemmatize_text(text):
    # Токенизация и лемматизация
    words = text.split()
    words = [lemmatizer.lemmatize(word) for word in words if word not in stop_words]
    return ' '.join(words)
````

## Векторизация изображений

Перейдём к векторизации изображений.

Самый примитивный способ — прочесть изображение и превратить полученную матрицу в вектор. Такой способ нам не подходит: длина векторов может быть сильно разной, так как размеры изображений разные. Поэтому стоит обратиться к свёрточным сетям: они позволяют "выделить" главные компоненты изображений. Как это сделать? Нужно выбрать какую-либо архитектуру, например ResNet-18, посмотреть на слои и исключить полносвязные слои, которые отвечают за конечное предсказание. При этом можно загрузить модель данной архитектуры, предварительно натренированную на датасете ImageNet.

````python
# Загружаем предобученный ResNet-18
model = models.resnet18(weights='IMAGENET1K_V1') #загружаю предварительно обученные веса нейронной сети на датасете ImageNet-1K
# Замораживаем веса
for param in model.parameters():
    param.requires_grad = False
model.eval()
# Удаляем последний слой fc и avgpool (можно оставить avgpool для стабилизации)
model = torch.nn.Sequential(
    *list(model.children())[:-1]
    )  # теперь выход: [batch, 512, 1, 1] если был avgpool


# Удаляем последний слой fc
model = torch.nn.Sequential(
    *list(model.children())[:-1], # теперь выход: [batch, 512, 1, 1]
    torch.nn.Flatten() # теперь выход: [batch, 512]   
    )  

# Трансформация под вход сети
transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])
````

````python
# Функция для векторизации всех изображений в папке
def vectorize_images_in_folder(folder_path):
    """
    Принимает путь к папке с изображениями.
    Возвращает словарь: {имя_файла: вектор_эмбеддинга}
    """
    folder_path = Path(folder_path)
    
    if not folder_path.exists():
        print(f" Папка не найдена: {folder_path}")
        return {}
    
    # Получаем все изображения в папке
    image_extensions = {'.jpg', '.jpeg', '.png', '.bmp', '.tiff'}
    image_files = [f for f in folder_path.iterdir() 
                   if f.suffix.lower() in image_extensions]
    
    if not image_files:
        print(f" В папке {folder_path} нет изображений")
        return {}
    
    embeddings_dict = {}
    
    with torch.no_grad():
        for img_path in image_files:
            try:
                # Загружаем и преобразуем изображение
                img = Image.open(img_path).convert('RGB')  # convert RGB на случай grayscale
                img_tensor = transform(img).unsqueeze(0)  # добавляем batch dimension
                
                # Получаем эмбеддинг
                embedding = model(img_tensor)
                
                # Сохраняем результат
                embeddings_dict[img_path.name] = embedding[0].cpu().numpy()
                #print(f" Обработано: {img_path.name}")
                
            except Exception as e:
                print(f" Ошибка при обработке {img_path.name}: {e}")
    
    return embeddings_dict
````

````python
#Векторизация текстов при помощи BERT
# Загрузка модели и токенизатора
tokenizer = transformers.BertTokenizer.from_pretrained('bert-base-cased')
model = transformers.BertModel.from_pretrained('bert-base-cased')
model.eval()
````

````python
#функции
def get_bert_embeddings(df, text_column='processed_query_text', max_length=128, batch_size=100):
    """
    Функция для получения BERT эмбеддингов из текстового столбца
    
    Args:
        df: DataFrame с текстами
        text_column: название столбца с текстами
        max_length: максимальная длина последовательности
        batch_size: размер батча
    
    Returns:
        numpy array с эмбеддингами
    """
    # Токенизация
    tokenized = df[text_column].apply(
        lambda x: tokenizer.encode(
            x, 
            add_special_tokens=True, 
            max_length=max_length, 
            truncation=True,
            padding=False  # Не паддим здесь, сделаем позже
        )
    )
    
    # Находим максимальную длину в этом датасете
    # проверяем на пустоту через len(tokenized)
    if len(tokenized) == 0:
        return np.array([])
    
    max_len = max(len(i) for i in tokenized)
    max_len = min(max_len, max_length)  # Ограничиваем max_length
    
    # Паддинг
    padded = np.array([i + [0]*(max_len - len(i)) for i in tokenized.values])
    attention_mask = np.where(padded != 0, 1, 0)
    
    # Извлечение эмбеддингов батчами
    embeddings = []
    
    for i in range(0, padded.shape[0], batch_size):
        batch = torch.LongTensor(padded[i:i+batch_size])
        mask_batch = torch.LongTensor(attention_mask[i:i+batch_size])
        
        with torch.no_grad():
            outputs = model(batch, attention_mask=mask_batch)
        
        # Берём [CLS] эмбеддинг
        cls_embeds = outputs.last_hidden_state[:, 0, :].numpy()
        embeddings.append(cls_embeds)
    
    return np.concatenate(embeddings, axis=0)
````

Молодец, что используешь эмбеддинг с CLS-токена — это стандартный подход для извлечения представления текста, особенно если модель была обучена на задаче NSP (Next Sentence Prediction), как это было с оригинальным BERT.

Однако стоит учитывать, что не все энкодерные архитектуры обучались на задаче NSP (например, DistilBERT).

На практике часто используют пулинг (например, средний) по всем выходным токенам (кроме паддингов), чтобы получить более общее представление текста. Многие исследователи отмечают, что при этом качество эмбеддингов оказывается выше. Если встретишь такой подход в чужом коде (или туториале), не удивляйся.

Вот пример реализации такого подхода:
    

```python
def preprocess_text_with_pooling(text):
    inputs = tokenizer(text, return_tensors='pt', truncation=True, padding='max_length', max_length=128).to(device)
    attention_mask = inputs['attention_mask']
    with torch.no_grad():
        outputs = distilbert_model(**inputs)

    last_hidden_state = outputs.last_hidden_state

    # Убираем влияние паддинг-токенов
    attention_mask = attention_mask.unsqueeze(-1)
    masked_hidden_state = last_hidden_state * attention_mask
    sum_hidden_states = masked_hidden_state.sum(dim=1)
    sum_mask = attention_mask.sum(dim=1)
    mean_hidden_state = sum_hidden_states / sum_mask

    return mean_hidden_state
```

Такой подход позволяет учесть всю доступную информацию в тексте, исключая влияние токенов-паддингов, что иногда даёт более качественные эмбеддинги.

В блоке использования **PCA (Метод главных компонент)** допущены две критические ошибки, нарушающие методологию машинного обучения:

1.  **Утечка данных (Data Leakage):** Обучение алгоритма PCA (метод `.fit()`) производится на всем массиве эмбеддингов до их разделения на обучающую и валидационную выборки. В результате информация о распределении данных из валидационной выборки «просачивается» в тренировочные признаки, что делает оценку качества модели необъективной. Необходимо выполнять `.fit()` строго на тренировочных данных, а к валидационным применять только `.transform()`.
2.  **Отсутствие масштабирования:** Перед применением PCA признаки **обязательно** должны быть приведены к одному масштабу (например, с помощью `StandardScaler`). PCA ищет направления максимальной дисперсии, и если признаки имеют разный разброс, алгоритм будет ориентироваться на масштаб чисел, а не на их значимость, что приводит к потере важной информации.

Нейронная сеть сама способна эффективно сжимать признаки в своих скрытых слоях, делая это осознанно (оптимизируя ошибку предсказания), в то время как PCA сжимает данные «вслепую», ориентируясь только на их разброс.

Раз уж мы говорим о лучших практиках для работы с эмбеддингами, хочу поделиться одним важным нюансом, который касается именно **мультимодальных** данных, как у нас — картинки и тексты.

Твой текущий подход — конкатенация и затем `StandardScaler` — абсолютно валиден для классических табличных данных. Но для эмбеддингов он может быть неоптимален. `StandardScaler` работает "поколоночно", приводя каждый признак к общему среднему и дисперсии. 


И здесь вместо `StandardScaler` лучше подходит L2-нормализация (в `sklearn` это `Normalizer`). Этот метод масштабирует каждый вектор (каждую строку целиком) к единичной длине. Он убирает влияние "магнитуды" вектора (которая часто является шумом) и оставляет только его **направление** в многомерном пространстве — а именно оно и несет основную семантическую информацию.
    
Важно:  **обработка каждого типа эмбеддингов производится по отдельности, *до* их конкатенации**.
    
Таким образом, более продвинутый пайплайн мог бы выглядеть так:
1.  Получаем матрицу эмбеддингов для изображений `X_img`.
2.  Получаем матрицу эмбеддингов для текстов `X_text`.
3.  Применяем `Normalizer()` к `X_img` и `X_text` **по отдельности**.
4.  Конкатенируем уже нормализованные матрицы.

Такой подход гарантирует, что и визуальные, и текстовые признаки вносят в модель сопоставимый вклад, не искажая при этом структуру каждого из исходных пространств. Это знание будет очень полезно тебе в будущих проектах.

При использовании `GridSearchCV` допущена критическая ошибка, приводящая к утечке данных (**data leakage**).

По умолчанию `GridSearchCV` использует стандартный `KFold`, который разбивает строки датасета случайно. В нашем проекте одно и то же изображение встречается в нескольких строках с разными описаниями. В результате случайного разбиения разные пары с одной и той же картинкой попадают одновременно и в обучение, и в валидацию. Модель «узнает» знакомые изображения вместо того, чтобы учиться обобщать, что делает полученные метрики бессмысленными.

**Как это исправить:**

1.  **Самый простой и эффективный путь:** Отказаться от кросс-валидации и использовать ту валидационную выборку, которую ты уже подготовила ранее (через `GroupShuffleSplit`). Это сэкономит время и вычислительные ресурсы. Если ты хочешь сохранить удобный интерфейс `GridSearchCV`, воспользуйся инструментом **`PredefinedSplit`**, чтобы зафиксировать твое разделение.
2.  **Если ты настаиваешь на кросс-валидации:** Используй **`GroupKFold`** в качестве параметра `cv`. При этом в метод `.fit()` обязательно нужно передать аргумент `groups` (столбец с именами файлов изображений), чтобы алгоритм не разделял одну группу (картинку) между фолдами.

#### Полносвязные нейронные сети

````python
print("\n" + "=" * 80)
print("ПОЛНОСВЯЗНЫЕ НЕЙРОННЫЕ СЕТИ")
print("=" * 80)

# Преобразуем данные в тензоры
X_train_tensor = torch.FloatTensor(X_train)
X_val_tensor = torch.FloatTensor(X_val)
y_train_tensor = torch.FloatTensor(y_train)
y_val_tensor = torch.FloatTensor(y_val)
y_binary_val = (y_val >= 0.5).astype(int)

# Создаем DataLoader
train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
val_dataset = TensorDataset(X_val_tensor, y_val_tensor)
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)  # Увеличил batch_size
val_loader = DataLoader(val_dataset, batch_size=64, shuffle=False)

class DeepNN(nn.Module):
    """Глубокая нейронная сеть с dropout и batch norm"""
    def __init__(self, input_dim, hidden_dims=[256, 128, 64, 32]):
        super(DeepNN, self).__init__()
        layers = []
        prev_dim = input_dim
        for hidden_dim in hidden_dims:
            layers.append(nn.Linear(prev_dim, hidden_dim))
            layers.append(nn.BatchNorm1d(hidden_dim))
            layers.append(nn.ReLU())
            layers.append(nn.Dropout(0.2))  # Уменьшил dropout
            prev_dim = hidden_dim
        layers.append(nn.Linear(prev_dim, 1))
        
        self.network = nn.Sequential(*layers)
    
    def forward(self, x):
        return self.network(x).squeeze()

class WideNN(nn.Module):
    """Широкая нейронная сеть с residual connections"""
    def __init__(self, input_dim, hidden_dim=512):
        super(WideNN, self).__init__()
        self.input_layer = nn.Linear(input_dim, hidden_dim)
        self.hidden1 = nn.Linear(hidden_dim, hidden_dim)
        self.hidden2 = nn.Linear(hidden_dim, hidden_dim // 2)
        self.hidden3 = nn.Linear(hidden_dim // 2, hidden_dim // 4)
        self.output_layer = nn.Linear(hidden_dim // 4, 1)
        
        self.bn1 = nn.BatchNorm1d(hidden_dim)
        self.bn2 = nn.BatchNorm1d(hidden_dim)
        self.bn3 = nn.BatchNorm1d(hidden_dim // 2)
        self.bn4 = nn.BatchNorm1d(hidden_dim // 4)
        
        self.dropout = nn.Dropout(0.3)
        
    def forward(self, x):
        # Входной слой
        x = self.input_layer(x)
        x = self.bn1(x)
        x = torch.relu(x)
        
        # Residual block 1
        residual = x
        x = self.hidden1(x)
        x = self.bn2(x)
        x = torch.relu(x)
        x = self.dropout(x)
        x = x + residual  # Residual connection
        
        # Hidden layer 2
        x = self.hidden2(x)
        x = self.bn3(x)
        x = torch.relu(x)
        x = self.dropout(x)
        
        # Hidden layer 3
        x = self.hidden3(x)
        x = self.bn4(x)
        x = torch.relu(x)
        x = self.dropout(x)
        
        # Output
        x = self.output_layer(x)
        return x.squeeze()

# Функция для обучения с улучшенными параметрами
def train_nn_model(model_class, model_name, input_dim, param_grid, num_epochs=150):
    """Обучает нейронную сеть с подбором гиперпараметров"""
    print(f"\n--- {model_name} ---")
    start_time = time.time()
    
    best_val_loss = float('inf')
    best_model = None
    best_params = None
    best_predictions = None
    
    total_combinations = len(param_grid['hidden_dims']) if 'hidden_dims' in param_grid else len(param_grid['hidden_dim'])
    print(f"Тестируем {total_combinations} комбинаций гиперпараметров...")
    
    for i in tqdm(range(total_combinations), desc=f"  Перебор параметров", leave=True):
        # Извлекаем параметры
        params = {}
        for key, values in param_grid.items():
            if isinstance(values, list):
                params[key] = values[i] if i < len(values) else values[0]
            else:
                params[key] = values
        
        # Создаем модель
        if 'hidden_dims' in params:
            model = model_class(input_dim, hidden_dims=params['hidden_dims'])
        else:
            hidden_dim = params.get('hidden_dim', 512)
            model = model_class(input_dim, hidden_dim=hidden_dim)
        
        # Оптимизатор с AdamW (лучше чем Adam)
        optimizer = optim.AdamW(model.parameters(), 
                                lr=params.get('lr', 0.001), 
                                weight_decay=params.get('weight_decay', 1e-4))
        
        # Cosine annealing scheduler
        scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=num_epochs)
        
        criterion = nn.BCEWithLogitsLoss()  # Лучше для бинарной классификации
        
        # Обучение
        best_epoch_val_loss = float('inf')
        patience_counter = 0
        early_stopping_patience = 30
        best_epoch = 0
        
        for epoch in range(num_epochs):
            # Train
            model.train()
            train_loss = 0
            for batch_X, batch_y in train_loader:
                optimizer.zero_grad()
                outputs = model(batch_X)
                loss = criterion(outputs, batch_y)
                loss.backward()
                torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)  # Gradient clipping
                optimizer.step()
                train_loss += loss.item()
            
            # Validation
            model.eval()
            val_loss = 0
            with torch.no_grad():
                for batch_X, batch_y in val_loader:
                    outputs = model(batch_X)
                    val_loss += criterion(outputs, batch_y).item()
            
            avg_train_loss = train_loss / len(train_loader)
            avg_val_loss = val_loss / len(val_loader)
            
            scheduler.step()
            
            # Early stopping
            if avg_val_loss < best_epoch_val_loss:
                best_epoch_val_loss = avg_val_loss
                patience_counter = 0
                best_epoch = epoch
            else:
                patience_counter += 1
                if patience_counter >= early_stopping_patience:
                    break
            
            # Печать прогресса каждые 10 эпох
            if (epoch + 1) % 20 == 0:
                print(f"    Эпоха {epoch+1}/{num_epochs} - Train Loss: {avg_train_loss:.4f}, Val Loss: {avg_val_loss:.4f}")
        
        print(f"    Лучшая эпоха: {best_epoch}, Val Loss: {best_epoch_val_loss:.4f}")
        
        # Сохраняем лучшую модель
        if best_epoch_val_loss < best_val_loss:
            best_val_loss = best_epoch_val_loss
            best_model = model
            best_params = params
            
            # Предсказания (логиты -> вероятности через sigmoid)
            model.eval()
            with torch.no_grad():
                logits = model(X_val_tensor)
                best_predictions = torch.sigmoid(logits).numpy()
    
    elapsed_time = time.time() - start_time
    print(f"\n Время обучения: {elapsed_time:.1f} сек ({elapsed_time/60:.1f} мин)")
    print(f" Лучшие параметры: {best_params}")
    
    # Метрики
    best_predictions_binary = (best_predictions >= 0.5).astype(int)
    
    rmse = np.sqrt(mean_squared_error(y_val, best_predictions))
    mae = mean_absolute_error(y_val, best_predictions)
    r2 = r2_score(y_val, best_predictions)
    roc_auc = roc_auc_score(y_binary_val, best_predictions)
    precision = precision_score(y_binary_val, best_predictions_binary)
    recall = recall_score(y_binary_val, best_predictions_binary)
    f1 = f1_score(y_binary_val, best_predictions_binary)
    
    print(f"\n Метрики на валидации:")
    print(f"  RMSE: {rmse:.4f}")
    print(f"  MAE: {mae:.4f}")
    print(f"  R²: {r2:.4f}")
    print(f"  Precision: {precision:.4f}")
    print(f"  Recall: {recall:.4f}")
    print(f"  F1-Score: {f1:.4f}")
    print(f"  ROC-AUC: {roc_auc:.4f}")
    
    return {
        'model': best_model,
        'predictions': best_predictions,
        'RMSE': rmse,
        'MAE': mae,
        'R2': r2,
        'Precision': precision,
        'Recall': recall,
        'F1-Score': f1,  
        'ROC-AUC': roc_auc,
        'params': best_params,
        'training_time': elapsed_time
    }

# Улучшенные гиперпараметры
deep_params_grid = {
    'hidden_dims': [
        [256, 128, 64, 32],
        [512, 256, 128, 64],
        [1024, 512, 256, 128],
        [512, 256, 128]
    ],
    'lr': [0.001, 0.0005],
    'weight_decay': [1e-5, 1e-4]
}

wide_params_grid = {
    'hidden_dim': [256, 512, 768],
    'lr': [0.001, 0.0005, 0.0001],
    'weight_decay': [1e-5, 1e-4, 5e-4]
}

# Обучаем модели
input_dim = X_train.shape[1]

nn_results = {}

# Глубокая NN
print("\n" + "=" * 60)
nn_results['DeepNN'] = train_nn_model(
    DeepNN, "Глубокая NN (4 скрытых слоя + BatchNorm)", 
    input_dim, deep_params_grid, num_epochs=150
)

# Широкая NN с residual connections
print("\n" + "=" * 60)
nn_results['WideNN'] = train_nn_model(
    WideNN, "Широкая NN (Residual connections)", 
    input_dim, wide_params_grid, num_epochs=150
)

# Вывод итогов
print("\n" + "=" * 80)
print(" СРАВНЕНИЕ МОДЕЛЕЙ")
print("=" * 80)

comparison_df = pd.DataFrame({
    model: {
        'RMSE': results['RMSE'],
        'MAE': results['MAE'],
        'R²': results['R2'],
        'F1-Score': results['F1-Score'],
        'ROC-AUC': results['ROC-AUC'],
        'Время (мин)': results['training_time']/60
    }
    for model, results in nn_results.items()
}).T

print(comparison_df.round(4))

print("\n" + "=" * 80)
print("ИТОГОВОЕ ВРЕМЯ ОБУЧЕНИЯ")
print("=" * 80)
for name, result in nn_results.items():
    print(f"{name}: {result['training_time']:.1f} сек ({result['training_time']/60:.1f} мин)")
````

````python
# Список слов, указывающих на детей (до 16 лет)
child_keywords = child_keywords #список определн выше в п.2

# Функция для проверки наличия детских слов в тексте
def contains_child_keywords(text):
    """
    Проверяет, содержит ли текст слова, указывающие на детей
    Возвращает True если содержит, иначе False
    """
    text_lower = text.lower()
    for keyword in child_keywords:
        if keyword in text_lower:
            return True
    return False
````

````python
# Пересоздаем BERT модель для поиска 
bert_tokenizer_search = transformers.BertTokenizer.from_pretrained('bert-base-cased')
bert_model_search = transformers.BertModel.from_pretrained('bert-base-cased')
bert_model_search.eval()
````

````python

# НОРМАЛИЗАЦИЯ ЭМБЕДДИНГОВ

def normalize_embedding(emb):
    """Нормализует эмбеддинг до единичной длины"""
    norm = np.linalg.norm(emb)
    return emb / norm if norm > 0 else emb

# Нормализация тестовых эмбеддингов изображений
print("\n Нормализация тестовых эмбеддингов...")
test_embeddings_normalized = {}
for img_name, emb in test_embeddings.items():
    test_embeddings_normalized[img_name] = normalize_embedding(emb)
print(f" Нормализовано {len(test_embeddings_normalized)} изображений")
````

````python
# ВЕКТОРИЗАЦИЯ ТЕКСТА

def get_text_embedding(text, tokenizer, model, max_length=128):
    """Получение BERT эмбеддинга для текста"""
    encoded = tokenizer.encode(
        text,
        add_special_tokens=True,
        max_length=max_length,
        truncation=True,
        padding='max_length',
        return_tensors='pt'
    )
    
    with torch.no_grad():
        outputs = model(encoded)
        embedding = outputs.last_hidden_state[:, 0, :].numpy()[0]
    
    return normalize_embedding(embedding)
````

````python
# ОСНОВНАЯ ФУНКЦИЯ ПОИСКА

def search_images(query_text):
    """Поиск изображений по текстовому запросу"""
    
    # Проверка на детский контент
    if contains_child_keywords(query_text):
        print("   ⛔ Запрос заблокирован (содержит упоминания детей)")
        return None
    
    # Получаем эмбеддинг текста
    text_emb = get_text_embedding(query_text, bert_tokenizer_search, bert_model_search)
    
    # Вычисляем оценки для всех изображений
    results = []
    for img_name, img_emb in test_embeddings_normalized.items():
        # Конкатенируем эмбеддинги
        combined = np.concatenate([text_emb, img_emb])
        
        # Преобразуем в тензор
        features_tensor = torch.FloatTensor(combined).reshape(1, -1)
        
        # Предсказание
        with torch.no_grad():
            logits = best_nn_model(features_tensor)
            score = torch.sigmoid(logits).item()
        
        results.append({
            'image': img_name,
            'score': score,
            'path': str(Path(path_to_image_folder_test) / img_name)
        })
    
    # Сортировка по убыванию score
    results.sort(key=lambda x: x['score'], reverse=True)
    return results
````

````python
# ОТОБРАЖЕНИЕ ИЗОБРАЖЕНИЙ


def show_images(results, top_k=3, query_text=""):
    """Отображает топ-K изображений"""
    if not results:
        return
    
    n = min(top_k, len(results))
    fig, axes = plt.subplots(1, n, figsize=(5*n, 5))
    if n == 1:
        axes = [axes]
    
    for i in range(n):
        try:
            img = Image.open(results[i]['path'])
            axes[i].imshow(img)
            axes[i].set_title(f"#{i+1} | Score: {results[i]['score']:.3f}", fontsize=10)
            
            # Красная рамка для низких оценок
            if results[i]['score'] < 0.5:
                for spine in axes[i].spines.values():
                    spine.set_color('red')
                    spine.set_linewidth(2)
        except:
            axes[i].text(0.5, 0.5, "Ошибка загрузки", ha='center', va='center')
        
        axes[i].axis('off')
    
    if query_text:
        short_text = query_text[:60] + "..." if len(query_text) > 60 else query_text
        fig.suptitle(f"Поиск: {short_text}", fontsize=12, fontweight='bold')
    
    plt.tight_layout()
    plt.show()

def compare_with_ground_truth(pred_path, gt_path, pred_score, gt_score, query_text):
    """Сравнивает предсказание с эталоном"""
    fig, axes = plt.subplots(1, 2, figsize=(10, 5))
    
    try:
        pred_img = Image.open(pred_path)
        axes[0].imshow(pred_img)
        axes[0].set_title(f"Топ-1 предсказание\nScore: {pred_score:.3f}", color='blue')
    except:
        axes[0].text(0.5, 0.5, "Ошибка", ha='center', va='center')
    axes[0].axis('off')
    
    try:
        gt_img = Image.open(gt_path)
        axes[1].imshow(gt_img)
        axes[1].set_title(f"Эталон\nScore: {gt_score:.3f}", color='green')
    except:
        axes[1].text(0.5, 0.5, "Ошибка", ha='center', va='center')
    axes[1].axis('off')
    
    short_text = query_text[:50] + "..." if len(query_text) > 50 else query_text
    fig.suptitle(f"Сравнение: {short_text}", fontsize=12)
    plt.tight_layout()
    plt.show()


````

````python
# ТЕСТИРОВАНИЕ

print("\n" + "=" * 80)
print(" ТЕСТИРОВАНИЕ МОДЕЛИ")
print("=" * 80)


# Выбираем случайные запросы
sample_queries = df_test_q.sample(n=N, random_state=RANDOM_STATE).reset_index(drop=True)
print(f"\n Выбрано {len(sample_queries)} запросов для тестирования")

# Тестирование
stats = {'total': 0, 'blocked': 0, 'rank1': 0, 'rank3': 0, 'rank5': 0, 'found': 0}

for idx, row in sample_queries.iterrows():
    query_text = row['query_text']
    query_id = row['query_id']
    ground_truth = row.get('image', None)
    
    print(f"\n{'='*60}")
    print(f"  ЗАПРОС {idx+1}/{len(sample_queries)}")
    print(f"   ID: {query_id}")
    print(f"   Текст: {query_text[:150]}")
    if ground_truth:
        print(f"    Эталон: {ground_truth}")
    print(f"{'='*60}")
    
    # Поиск
    results = search_images(query_text)
    
    if results is None:
        stats['total'] += 1
        stats['blocked'] += 1
        continue
    
    stats['total'] += 1
    
    # Вывод топ-3
    print(f"\n    ТОП-3 РЕЗУЛЬТАТА:")
    for i, res in enumerate(results[:3]):
        status = "✅" if res['score'] >= 0.5 else "❌"
        print(f"      {i+1}. {status} {res['image']} ({res['score']:.4f})")
    
    # Отображение
    print(f"\n    ВИЗУАЛИЗАЦИЯ:")
    show_images(results, top_k=3, query_text=query_text)
    
    # Проверка эталона
    if ground_truth:
        rank = None
        for i, res in enumerate(results):
            if res['image'] == ground_truth:
                rank = i + 1
                break
        
        if rank:
            stats['found'] += 1
            if rank == 1:
                stats['rank1'] += 1
                stats['rank3'] += 1
                stats['rank5'] += 1
                print(f"\n   ✅ ОТЛИЧНО! Эталон на позиции #{rank}")
            elif rank <= 3:
                stats['rank3'] += 1
                stats['rank5'] += 1
                print(f"\n   👍 ХОРОШО! Эталон на позиции #{rank}")
            elif rank <= 5:
                stats['rank5'] += 1
                print(f"\n   👌 УДОВЛЕТВОРИТЕЛЬНО! Эталон на позиции #{rank}")
            else:
                print(f"\n   ⚠️ Эталон на позиции #{rank}")
            
            # Сравнение если эталон не первый
            if rank != 1 and rank <= 10:
                gt_path = str(Path(path_to_image_folder_test) / ground_truth)
                if Path(gt_path).exists():
                    compare_with_ground_truth(
                        results[0]['path'], gt_path,
                        results[0]['score'], results[rank-1]['score'],
                        query_text
                    )
        else:
            print(f"\n   ❌ Эталон НЕ НАЙДЕН")

````

<p style="font-size: 1.4em; font-weight: bold; margin-top: 30px; margin-bottom: 10px;">От стажера до мастера: практическое руководство по работе с AI</p>
<p>Итак, как использовать AI, чтобы он усиливал вас, а не заменял?</p>
<ol>
    <li><strong>Освойте новый навык — промпт-инжиниринг.</strong> Умение правильно ставить задачи AI — это ключевая компетенция. Будьте конкретны, предоставляйте контекст, примеры, описывайте желаемый формат вывода. Чем лучше ваш запрос, тем качественнее результат.</li>
